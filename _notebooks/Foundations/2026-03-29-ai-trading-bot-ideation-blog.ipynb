The bot combines both to make a short term prediction, basically a guess about tomorrow, not next year. On top of that, we're not building it once and calling it done. We're using an approach called **DBR**, which stands for Design, Build, Refine. Build something small, test if it actually works, then improve it. Repeat.

The bot isn't going to live in isolation either. We're wiring it into two other projects, **Fortune Finders** and **Wand**, so it becomes something people can actually play with, not just a script that runs quietly in the background.

<h2 style="border-left:4px solid #DDAA4E; padding-left:12px;">A Small Peek at the Logic</h2>

The real model is more involved than this, but here's the basic idea behind sentiment scoring, stripped down to something easy to read. Given a headline, count the positive and negative words in it and see which side wins.

```python
# CODE_RUNNER: Add your own headline to the list and see how it scores.

good_words = ["beats", "growth", "record", "strong", "upgrade"]
bad_words = ["misses", "drop", "weak", "lawsuit", "downgrade"]

def score_headline(headline):
    words = headline.lower().split()
    positive = sum(1 for w in words if w in good_words)
    negative = sum(1 for w in words if w in bad_words)
    if positive > negative:
        return "positive"
    elif negative > positive:
        return "negative"
    else:
        return "neutral"

headlines = [
    "Company beats earnings expectations with strong growth",
    "Stock drops after company misses on revenue",
]

for h in headlines:
    print(h, "->", score_headline(h))
```

That's the whole idea in miniature. The real version we're building uses a proper sentiment model instead of a word list, but the concept is the same, read the headline, decide if it leans good or bad, and let that feed into the prediction.
